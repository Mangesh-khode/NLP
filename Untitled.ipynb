```python
import math
from collections import Counter

# List of documents
documents = [
    "The quick brown fox jumps over the lazy dog",
    "The lazy dog likes to sleep all day",
    "The brown fox prefers to eat cheese",
    "The red fox jumps over the brown fox",
    "The brown dog chases the fox"
]

# Preprocessing: Convert documents to lowercase and split into words
processed_docs = [doc.lower().split() for doc in documents]

# Terms we are interested in
terms = ["brown", "fox", "dog", "cheese", "jump"]

# Function to calculate term frequency (TF)
def compute_tf(term, document):
    term_count = document.count(term)
    total_terms = len(document)
    return term_count / total_terms

# Function to calculate inverse document frequency (IDF)
def compute_idf(term, documents):
    num_docs_containing_term = sum(1 for doc in documents if term in doc)
    total_docs = len(documents)
    if num_docs_containing_term == 0:
        return 1e-10  # or handle it with a different approach, like skipping TF-IDF calculation
    return math.log(total_docs / num_docs_containing_term)

# Function to calculate TF-IDF
def compute_tf_idf(term, document, documents):
    tf = compute_tf(term, document)
    idf = compute_idf(term, documents)
    return tf * idf

# Calculate and print TF-IDF for each term in each document
for term in terms:
    print(f"\nTF-IDF for term '{term}':")
    for i, doc in enumerate(processed_docs):
        tf_idf_value = compute_tf_idf(term, doc, processed_docs)
        print(f"Document {i+1}: {tf_idf_value:.4f}")
```
