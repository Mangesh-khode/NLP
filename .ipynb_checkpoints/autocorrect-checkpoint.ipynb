```python
def get_probs(word_count_dict):
    probs={}
    m=sum(word_count_dict.values())
    for key in word_count_dict.keys():
        probs[key]=word_count_dict[key]/m
    return probs
```

```python
def delete_letter(word, verbose=False):
    # Generate list of tuples where each tuple contains the word split into two parts
    split_1 = [(word[:c], word[c:]) for c in range(len(word))]
    
    # Create list of words obtained by deleting each letter
    delete_1 = [a + b[1:] for a, b in split_1]
    
    # Print detailed information if verbose is True
    if verbose:
        print(f"Input word: {word}")
        print(f"Split_1: {split_1}")
        print(f"Delete_1: {delete_1}")
    
    return delete_1

# Call the function with verbosity enabled
delete_word_1 = delete_letter(word="cans", verbose=True)
```

```python
def switch_letter(word, verbose=False):
    def swap(c, i, j):
        c = list(c)
        c[i], c[j] = c[j], c[i]
        return ''.join(c)  # Join the list of characters back into a string
    
    switch_1 = []
    split_1 = []
    len_word = len(word)
    
    # Create a list of tuples splitting the word at each index
    for c in range(len_word):
        split_1.append((word[:c], word[c:]))
    
    # Generate list of words by swapping adjacent letters
    switch_1 = [swap(a + b[:2], 0, 1) + b[2:] for a, b in split_1 if len(b) >= 2]
    
    if verbose:
        print(f"Input word: {word}")
        print(f"Split_1: {split_1}")
        print(f"Switch_1: {switch_1}")
    
    return switch_1

# Example usage
switch_word_1 = switch_letter(word="eta", verbose=True)
```

```python
def replace_letter(word, verbose=False):
    letters = "abcdefghijklmnopqrstuvwxyz"  # Corrected variable name
    replace_1 = []
    split_1 = []
    
    # Split the word at each index
    for c in range(len(word)):
        split_1.append((word[:c], word[c:]))
    
    # Generate words by replacing each letter with every letter in 'letters'
    replace_1 = [a + l + (b[1:] if len(b) > 1 else '') for a, b in split_1 if b for l in letters]
    
    # Remove the original word and sort the list
    replace_set = set(replace_1)
    replace_set.discard(word)  # Use discard to avoid KeyError if the word is not in the set
    replace_1 = sorted(list(replace_set))
    
    if verbose:
        print(f"input word={word} \nsplit_1={split_1} \nreplace_1={replace_1}")
        
        print(len(replace_1))
    
    return replace_1  # Corrected return variable name

# Example usage
replace_word_1 = replace_letter(word="can", verbose=True)
```

```python
def insert_letter(word, verbose=False):
    letters = "abcdefghijklmnopqrstuvwxyz" 
    insert_l = []
    split_l = []  # Corrected name from split_1 to split_l
    
    for c in range(len(word) + 1):
        split_l.append((word[0:c], word[c:]))  # Fixed slicing issue
    
    insert_l = [a + l + b for a, b in split_l for l in letters]
    
    if verbose:
        print(f"Insertions for word '{word}' \nSplit: {split_l} \nInsertions: {insert_l}")  # Fixed print statement
    
    return insert_l
```

```python
def edit_one_letter(word, allow_switches=True):
    edit_one_set = set()
    edit_one_set.update(delete_letter(word))
    
    if allow_switches:
        edit_one_set.update(switch_letter(word))
        
    edit_one_set.update(replace_letter(word))
    edit_one_set.update(insert_letter(word))
    
    return edit_one_set

imp_word = "at"
imp_edit_one_set = edit_one_letter(imp_word)
imp_edit_one_1 = sorted(list(imp_edit_one_set))

print(f"Input word: {imp_word} \nEdit one set: {imp_edit_one_1}\n")
print(type(imp_edit_one_set))  # This prints the type of the set
print(len(imp_edit_one_set))   # Corrected to get the length of the set, not the function
```

```python
def edit_two_letters(word, allow_switches=True):
    edit_two = set()
    edit_one_set = edit_one_letter(word, allow_switches=allow_switches)  # Get words with edit distance of 1
    for w in edit_one_set:
        if w:
            edit_two.update(edit_one_letter(w, allow_switches=allow_switches))  # Get edit distance of 1 for each word in edit_one_set
    return edit_two

tmp_edit_two_set = edit_two_letters("a")
tmp_edit_two_1 = sorted(list(tmp_edit_two_set))
print(f"Number of strings with edit distance of two: {len(tmp_edit_two_1)}")
print(f"First 10 strings: {tmp_edit_two_1[:10]}")
print(f"Last 10 strings: {tmp_edit_two_1[-10:]}")
print(f"The data type of the return object should be a set: {type(tmp_edit_two_set)}")
print(f"Number of strings with edit distance of two for 'at': {len(edit_two_letters('at'))}")
```

```python
def get_correction(word ,probs, vocab,n=2,verbose=False):
    suggestion=[]
    n_best=[]
    suggestions=list((word in vocab and word) or edit_one_letter(word).intersection(vocab) or edit_two_letter(word).intersection(vocab))
    n_best=[[s,probs[s]] for s in list(reversed(suggestions))]
    if verbose:
        print("suggestions=",suggestions)
    return n_best
```

```python
my_word = 'dys'
tmp_correction = get_correction(my_word, probs, vocab, 2, verbose=False)
for i, word_prob in enumerate(tmp_correction):
    print(f"word{i}: {word_prob[0]}, probability {word_prob[1]:.6f}")
print(type(tmp_correction))
```

```python
import numpy as np

def min_edit_distance(source, target, ins_cost=1, del_cost=1, rep_cost=2):
    m = len(source)
    n = len(target)
    d = np.zeros((m + 1, n + 1), dtype=int)
    
    for row in range(1, m + 1):
        d[row, 0] = d[row - 1, 0] + del_cost
    
    for col in range(1, n + 1):
        d[0, col] = d[0, col - 1] + ins_cost
    
    for row in range(1, m + 1):
        for col in range(1, n + 1):
            r_cost = rep_cost
            if source[row - 1] == target[col - 1]:
                r_cost = 0
            d[row, col] = min(
                d[row - 1, col] + del_cost,       # Cost for deletion
                d[row, col - 1] + ins_cost,       # Cost for insertion
                d[row - 1, col - 1] + r_cost      # Cost for replacement
            )
    
    return d, d[m, n]
```

```python
import pandas as pd
source='play'
target='stay'
matrix, min_edits=min_edit_distance(source,target)
print("min edit distance", min_edits,"\n")
idx=list('#'+source)
cols=list('#'+target)
df=pd.DataFrame(matrix,index=idx, columns=cols)
print(df)
```
