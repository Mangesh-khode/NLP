```python
def fac(n):
    return 1 if(n==1 or n==0) else (n*fac(n-1))
num=6
print(fac(num))
```

```python
def fibo(n):
    if n<0:
        return 0
    elif n==0:
        return 0
    elif n==1 or n==2:
        return 1
    else :
        return fibo(n-1)+fibo(n-2)
print(fibo(9))
```

```python
def large(numbers):
    largest=numbers[0]
    for num in numbers:
        if num>largest:
            largest=num
    return largest
n=[10, 5, 8, 20, 3]
print(large(n))
```

```python
def ele(list1,list2):
    comm=[]
    for i in list1:
        if i in list2:
            comm.append(i)
    return comm
list_a = [1, 2, 3, 4, 5]
list_b = [4, 5, 6, 7, 8]
common = ele(list_a, list_b)
print(common)
    
```

```python
def pali(string):
    reversed_string=string[::-1]
    return string==reversed_string

word="madam"
if pali(word):
    print("palindrom")
else:
    print("not pallindrom")
```

```python
def largest(numbers):
    largest=numbers[0]
    for num in numbers:
        if num>largest:
            largest=num
    return largest
```
