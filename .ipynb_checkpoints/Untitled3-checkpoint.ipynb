```python
import requests
from bs4 import BeautifulSoup
import csv

def fetch_reviews(imdb_url, output_file):
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3'
    }
    # Fetch the content of the movie's review page with headers
    response = requests.get(imdb_url, headers=headers)
    if response.status_code != 200:
        print("Failed to retrieve page. Status code:", response.status_code)
        return

    # Parse the content with BeautifulSoup
    soup = BeautifulSoup(response.content, 'html.parser')

    # Find all review text sections
    reviews = soup.find_all('div', class_='text show-more__control')

    if not reviews:
        print("No reviews found or unable to parse reviews.")
        return

    # Open the CSV file for writing
    with open(output_file, 'w', newline='', encoding='utf-8') as csvfile:
        writer = csv.writer(csvfile)
        writer.writerow(['Review'])  # Writing header

        # Write each review to the CSV file
        for i, review in enumerate(reviews):
            review_text = review.get_text(strip=True)
            writer.writerow([review_text])

    print(f"Saved {len(reviews)} reviews to {output_file}")

# Replace this URL with the IMDb URL of the Punjabi movie's reviews page
imdb_url = 'https://www.imdb.com/title/tt16211184/reviews/?ref_=tt_ql_2'  # Example URL

# Output CSV file name
output_file ='mrajaney.csv'

# Fetch and save the reviews to CSV
fetch_reviews(imdb_url, output_file)
```
