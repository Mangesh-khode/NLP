```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('NationalNames.csv')
data = np.array(data['Name'][:10000]).reshape(-1,1)
data = [x.lower() for x in data[:,0]]
data = np.array(data).reshape(-1,1)

print("Data Shape ={}".format(data.shape))
print()
print("Lets see some names: ")
print(data[1:10])

transform_data = np.copy(data)
max_length = 0
for index in range(len(data)):
    max_length = max(max_length, len(data[index,0]))

for index in range(len(data)):
    length = (max_length - len(data[index,0]))
    string = '.'*length
    transform_data[index,0] = ''.join([transform_data[index,0], string])


print(transform_data)
```

```python
print(transform_data[1:10])
```

```python
vocab = []  # Initialize an empty list for vocab
for name in transform_data[:, 0]:
    vocab.extend(list(name))

vocab = set(vocab)  # Convert list to set to get unique characters
vocab_size = len(vocab)
print("Vocab size = {}".format(vocab_size))
print("Vocab      = {}".format(vocab))

# Dictionaries for character to ID and vice versa
char_id = dict()
id_char = dict()

for i, char in enumerate(vocab):
    char_id[char] = i
    id_char[i] = char

print('a-{},22-{}'.format(char_id.get('a', 'not found'), id_char.get(22, 'not found')))

train_dataset = []
batch_size = 5
transforms_data=transform_data[1:10]
# Loop through data to create batches
for i in range(len(transforms_data) - batch_size + 1):
    start = i * batch_size
    end = start + batch_size
    batch_data = transforms_data[start:end]

    if len(batch_data) != batch_size:
        break

    char_list = []
    
    # Iterate through characters in the first name of the batch
    for k in range(len(batch_data[0][0])):
        batch_dataset = np.zeros([batch_size, len(vocab)])  # Initialize batch matrix

        for j in range(batch_size):
            name = batch_data[j][0]
            char_index = char_id.get(name[k], -1)  # Use .get() to avoid errors if char is missing
            if char_index != -1:
                batch_dataset[j, char_index] = 1.0  # One-hot encoding

        char_list.append(batch_dataset)
    
    train_dataset.append(char_list)
```

```python
def sigmoid(X):
    return 1/(1+np.exp(-X))


def tanh_activation(X):
    return np.tanh(X)

def softmax(X):
    exp_X=np.exp(X)
    exp_X_sum=np.sum(exp_X,axis=1).reshape(-1,1)

    exp_X=exp_X/exp_X_sum
    return exp_X

def tanh_derivative(X):
    return 1-(X**2)
    
```

```python
def initialize_parameters():
    mean = 0
    std = 0.01

    # Define dimensions of each gate weight matrix
    forget_gate_weights = np.random.normal(mean, std, (input_units + hidden_units, hidden_units))
    input_gate_weights = np.random.normal(mean, std, (input_units + hidden_units, hidden_units))
    output_gate_weights = np.random.normal(mean, std, (input_units + hidden_units, hidden_units))
    get_gate_weights = np.random.normal(mean, std, (input_units + hidden_units, hidden_units))
    hidden_output_weights = np.random.normal(mean, std, (input_units + hidden_units, hidden_units))

    # Initialize parameters dictionary
    parameters = dict()
    parameters['fgw'] = forget_gate_weights
    parameters['igw'] = input_gate_weights
    parameters['ogw'] = output_gate_weights
    parameters['ggw'] = get_gate_weights
    parameters['how'] = hidden_output_weights

    return parameters
```
