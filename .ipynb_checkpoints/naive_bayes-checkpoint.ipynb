```python
import nltk
from os import getcwd
import numpy as np 
import pandas as pd 
from nltk.corpus import twitter_samples  
from nltk.corpus import stopwords

nltk.download('twitter_samples')
nltk.download('stopwords')

filepath = getcwd()
nltk.data.path.append(filepath)

all_positive_tweets = twitter_samples.strings("positive_tweets.json")
all_negative_tweets = twitter_samples.strings("negative_tweets.json")  

test_pos = all_positive_tweets[4000:]
train_pos = all_positive_tweets[:4000]
test_neg = all_negative_tweets[4000:]
train_neg = all_negative_tweets[:4000]

train_x = train_pos + train_neg
test_x = test_pos + test_neg

train_y = np.append(np.ones((len(train_pos), 1)), np.zeros((len(train_neg), 1)), axis=0)
test_y = np.append(np.ones((len(test_pos), 1)), np.zeros((len(test_neg), 1)), axis=0)

print("train_y.shape = " + str(train_y.shape))
print("test_y.shape = " + str(test_y.shape))

# Example of a simple `process_tweet` function
def process_tweet(tweet):
    # Tokenization and removal of stopwords as an example
    stop_words = set(stopwords.words('english'))
    tokens = nltk.word_tokenize(tweet)
    cleaned_tokens = [word for word in tokens if word.lower() not in stop_words and word.isalpha()]
    return cleaned_tokens

# Example of a simple `build_freqs` function
def build_freqs(tweets, ys):
    freqs = {}
    for y, tweet in zip(ys, tweets):
        for word in process_tweet(tweet):
            pair = (word, y[0])
            if pair in freqs:
                freqs[pair] += 1
            else:
                freqs[pair] = 1
    return freqs

freqs = build_freqs(train_x, train_y)

print("type(freqs) = " + str(type(freqs)))
print("len(freqs) = " + str(len(freqs.keys())))

print("This is an example of positive tweets:\n", train_x[0])
print("\nThis is an example of the processed version of the tweet:\n", process_tweet(train_x[0]))

def sigmoid(z):
    h = 1 / (1 + np.exp(-z))
    return h

if sigmoid(0) == 0.5:
    print("Yes")
else:
    print("Oops")
```

```python
def gradientDescent(x, y, theta, alpha, num_iters):
    m = x.shape[0]  # number of training examples
    for i in range(num_iters):
        z = np.dot(x, theta)
        h = sigmoid(z)
        
        # Compute the cost function
        j = -1. / m * (np.dot(y.T, np.log(h)) + np.dot((1 - y).T, np.log(1 - h)))
        
        # Update theta
        theta -= alpha * (1/m) * np.dot(x.T, (h - y))
    
#     j=float[j]
    return theta, j
```

```python
def extract_features(tweet,freqs):
    word1=process_tweet(tweet)
    x=np.zeros((1,3))
    x[0,0]=1
    for word in word1:
        x[0,1]+=freqs.get((word,1.0),0)
        x[0,2]+=freqs.get((word,0,0),0)
        
    assert(x.shape==(1,3))
    return x
```

```python
import numpy as np

# Assuming extract_features and gradientDescent functions are defined

# Initialize feature matrix x
x = np.zeros((len(train_x), 3))

# Populate x with features extracted from train_x
for i in range(len(train_x)):
    x[i, :] = extract_features(train_x[i], freqs)  # Fixed indexing from 1 to i

# Define y
y = train_y

# Perform gradient descent to find optimal theta
theta, j = gradientDescent(x, y, np.zeros((3, 1)), 1e-9, 1500)

print(j)
print(theta)
# # Print the cost after training
# print(f"The cost after training is: {j:.8f}")

# # Print the resultant vector of weights (theta)
# print(f"The resultant vector of weights (theta) is: {[round(t, 8) for t in np.squeeze(theta)]}")
```

```python
def predict_tweet(tweet,freqs,theta):
    x=extract_features(tweet,freqs)
    y_pred=sigmoid(np.dot(x,theta))
    return y_pred
```

```python
for tweet in ["i am happy","i am sad","this movie should have great",'great','great great great']:
    print('%s->%f'%(tweet,predict_tweet(tweet,freqs,theta)))
```

```python
import numpy as np

def test_log(test_x, test_y, freqs, theta):
    y_hat = []
    for tweet in test_x:
        y_pred = predict_tweet(tweet, freqs, theta)
        if y_pred > 0.5:
            y_hat.append(1)
        else:
            y_hat.append(0)
    
    accuracy = (np.array(y_hat) == np.squeeze(test_y)).sum() / len(test_x)
    return accuracy
```

```python
imp_accuracy=test_log(test_x,test_y,freqs,theta)
print(imp_accuracy)
```

```python
def count_tweets(result,tweet,ys):
    for y , tweet in zip(ys,tweet):
        for word in process_tweet(tweet):
            pair = (word,y)
            if pair in result:
                result[pair]+=1
            else:
                result[pair]=1
    return result
```

```python
import numpy as np

def lookup(freqs, word, label):
    n = 0
    pair = (word, label)
    if pair in freqs:
        n = freqs[pair]  # Corrected this line to use freqs[pair]
    return n

def train_naive_bayes(freqs, train_x, train_y):
    vocab = set([pair[0] for pair in freqs.keys()])  # Unique words in the vocabulary
    y = len(vocab)
    
    N_pos = N_neg = V_pos = V_neg = 0
    for pair in freqs.keys():
        if pair[1] > 0:
            V_pos += 1
            N_pos += freqs[pair]
        else:
            V_neg += 1
            N_neg += freqs[pair]
    
    D = len(train_y)
    D_pos = len([x for x in train_y if x > 0])
    D_neg = len([x for x in train_y if x <= 0])
    
    logprior = np.log(D_pos) - np.log(D_neg)
    
    loglikelihood = {}
    for word in vocab:
        freq_pos = lookup(freqs, word, 1)
        freq_neg = lookup(freqs, word, 0)
        p_w_pos = (freq_pos + 1) / (N_pos + y)
        p_w_neg = (freq_neg + 1) / (N_neg + y)
        loglikelihood[word] = np.log(p_w_pos / p_w_neg)
        
    return logprior, loglikelihood
```

```python
#Example usage:
logprior, loglikelihood = train_naive_bayes(freqs, train_x, train_y)
print(logprior)
print(len(loglikelihood))
```
