```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import nltk
from nltk.corpus import stopwords
import string

# Download stopwords if not already downloaded
nltk.download('stopwords')

# Load Marathi stopwords
stop_words = set(stopwords.words('marathi'))

# Function for preprocessing reviews
def preprocess_text(text):
    # Convert to lowercase
    text = text.lower()
    # Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))
    # Tokenize and remove stop words
    text_tokens = text.split()
    text_tokens = [word for word in text_tokens if word not in stop_words]
    # Join tokens back into string
    return ' '.join(text_tokens)

# Load the dataset (Assuming columns are 'review' and 'label')
df = pd.read_csv('path_to_your_csv_file.csv')

# Preprocess the review text
df['cleaned_review'] = df['review'].apply(preprocess_text)

# Initialize the TF-IDF Vectorizer
vectorizer = TfidfVectorizer()

# Generate TF-IDF vectors for the reviews
tfidf_matrix = vectorizer.fit_transform(df['cleaned_review'])

# Compute cosine similarity between all documents
cosine_sim_matrix = cosine_similarity(tfidf_matrix)

# Function to find most similar reviews based on cosine similarity
def find_similar_reviews(index, cosine_sim_matrix, df, top_n=5):
    # Get the similarity scores for the given review
    similar_indices = cosine_sim_matrix[index].argsort()[:-top_n-1:-1]
    
    # Retrieve the reviews and their similarity scores
    similar_reviews = [(df.iloc[i]['review'], cosine_sim_matrix[index][i]) for i in similar_indices]
    
    return similar_reviews

# Example: Get top 5 similar reviews for the first review in the dataset
index = 0  # Change the index based on the review you want to compare
similar_reviews = find_similar_reviews(index, cosine_sim_matrix, df, top_n=5)

# Display the most similar reviews
for i, (review, score) in enumerate(similar_reviews):
    print(f"Review {i+1} (Similarity Score: {score}):\n{review}\n")
```
