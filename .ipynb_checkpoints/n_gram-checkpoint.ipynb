```python
def count_words(tokenized_sentences):
    word_count = {}  # Use a dictionary to store word counts

    # Iterate over each sentence
    for sentence in tokenized_sentences:
        # Iterate over each token (word) in the sentence
        for token in sentence:
            # If the token is not already in the word_count dictionary, initialize it to 1
            if token not in word_count:
                word_count[token] = 1
            else:
                # If the token is already in the dictionary, increment its count
                word_count[token] += 1

    return word_count

# Sample tokenized sentences
tokenized_sentences = [['sky', 'is', 'blue', '.'],
                       ['leaves', 'are', 'green', '.'],
                       ['roses', 'are', 'red']]

# Call the function and print the result
word_counts = count_words(tokenized_sentences)
print(word_counts)
```

```python
def get_word_with_nplus_frequency(tokenized_sentences, count_threshold):
    closed_vocab = []
    
    # Count the words in the tokenized sentences
    word_counts = count_words(tokenized_sentences)
    
    # Iterate through the word counts
    for word, cnt in word_counts.items():
        # Check if the word count is greater than or equal to the threshold
        if cnt >= count_threshold:
            closed_vocab.append(word)  # Add the word to the closed vocabulary

    return closed_vocab

# Sample tokenized sentences
tokenized_sentences = [['sky', 'is', 'blue', '.'],
                       ['leaves', 'are', 'green', '.'],
                       ['roses', 'are', 'red']]

# Define the count threshold
count_threshold = 2

# Call the function and print the result
tmp_closed_vocab = get_word_with_nplus_frequency(tokenized_sentences, count_threshold)
print("closed vocab")
print(tmp_closed_vocab)
```

```python
def replace_oov_word(tokenized_sentences, vocabulary, unknown_token='<unk>'):
    # List to hold the new tokenized sentences after OOV replacement
    replaced_tokenized_sentences = []

    # Iterate through each sentence in the tokenized sentences
    for sentence in tokenized_sentences:
        # List to hold the tokens of the current sentence after OOV replacement
        replaced_sentence = []
        
        # Iterate through each token (word) in the sentence
        for token in sentence:
            # Check if the token is in the vocabulary
            if token in vocabulary:
                replaced_sentence.append(token)  # Keep the token if it's in the vocabulary
            else:
                replaced_sentence.append(unknown_token)  # Replace OOV word with unknown token
        
        # Append the replaced sentence to the list of replaced sentences
        replaced_tokenized_sentences.append(replaced_sentence)
    
    return replaced_tokenized_sentences

# Corrected variable names and function call
tokenized_sentences = [["dog", "run"], ["cats", "sleep"]]
vocabulary = ["dogs", "sleep"]

# Call the function with corrected names
replaced_tokenized_sentences = replace_oov_word(tokenized_sentences, vocabulary)

# Display results
print(replaced_tokenized_sentences)
print(vocabulary)
```

```python
def preprocess_data(train_data, test_data, count_threshold):
    # Get vocabulary based on words with frequency greater than or equal to the count_threshold
    vocabulary = get_word_with_nplus_frequency(train_data, count_threshold)
    
    # Replace out-of-vocabulary words in train and test data
    train_data_replaced = replace_oov_word(train_data, vocabulary)
    test_data_replaced = replace_oov_word(test_data, vocabulary)
    
    # Return the preprocessed train data, test data, and the vocabulary
    return train_data_replaced, test_data_replaced, vocabulary
```

```python
def estimate_pro(word, previous_ngram, ngram_counts, n_plus_gram_counts, vocab_size, k=1.0):
    previous_ngram = tuple(previous_ngram)
    
    # Get count of the previous n-gram
    previous_ngram_count = ngram_counts[previous_ngram] if previous_ngram in ngram_counts else 0
    denominator = previous_ngram_count + k * vocab_size
    
    # Create the (n+1)-gram by adding the word to the previous n-gram
    n_plus_gram = previous_ngram + (word,)
    n_plus_ngram_count = n_plus_gram_counts[n_plus_gram] if n_plus_gram in n_plus_gram_counts else 0
    
    numerator = n_plus_ngram_count + k
    
    # Calculate probability
    probability = numerator / denominator
    
    return probability

# Example sentences
sentences = [['i', 'like', 'a', 'cat'],
             ['this', 'dog', 'is', 'like', 'a', 'cat']]

# Unique words in the sentences
unique_words = list(set(sentences[0] + sentences[1]))
print("lenght of unique_words:",len(unique_words))
# Get n-gram counts
unigram_counts = count_ngrams(sentences, 1)  # Unigram counts
print(unigram_counts)
bigram_counts = count_ngrams(sentences, 2)   # Bigram counts
print(bigram_counts)
# Example usage of estimate_pro
tmp_prob = estimate_pro("cat", "a", unigram_counts, bigram_counts, len(unique_words), k=1)
print(tmp_prob)
```

```python
import math
import random
import numpy as np
import pandas as pd
import nltk

nltk.data.path.append('.')  # Ensures the nltk data is searched in the current directory

# Download the punkt tokenizer model for tokenizing sentences/words
nltk.download('punkt')

# Reading the file
with open("en_US.twitter.txt", "r", encoding="utf-8") as f:
    data = f.read()

print("Data type: ", type(data))
print("Number of letters: ", len(data))

print("First 300 letters:")
print(data[0:300])

print("Last 300 letters:")
print(data[-300:])

# Function to split data into sentences based on newlines
def split_to_sentences(data):
    sentences = data.split('\n')  # Corrected: splitting by newlines
    sentences = [s.strip() for s in sentences]  # Stripping leading/trailing spaces
    sentences = [s for s in sentences if len(s) > 0]  # Removing empty sentences
    return sentences

# Test the sentence splitting function
x = """I have a pen. I have an apple. \nAh\n Apple pen.\n"""
print("\nOriginal text:\n", x)
sentences = split_to_sentences(x)
print("\nSplit sentences:\n", sentences)

# Tokenize each sentence into words
def tokenize_sentence(sentences):
    tokenized_sentences = []
    for sentence in sentences:
        sentence = sentence.lower()  # Convert the sentence to lowercase
        tokenized = nltk.word_tokenize(sentence)  # Tokenize the sentence into words
        tokenized_sentences.append(tokenized)  # Add the tokenized sentence to the list
        
    return tokenized_sentences

# Test the tokenization function
test_sentences = ['Sky is blue.', 'Leaves are green.', 'Roses are red.']
tokenized = tokenize_sentence(test_sentences)
print("\nTokenized sentences:\n", tokenized)

# Function to get tokenized data
def get_tokenized_data(data):
    sentences = split_to_sentences(data)  # Split data into sentences
    tokenized_sentences = tokenize_sentence(sentences)  # Tokenize sentences
    return tokenized_sentences

# Test string (this is your data)
x = "sky is blue.\nleaves are green.\nroses are red."

# Tokenizing the data
tokenized_data = get_tokenized_data(x)

# Shuffle the tokenized data with a seed for reproducibility
random.seed(87)
random.shuffle(tokenized_data)

# Split into train and test data (80% train, 20% test)
train_size = int(len(tokenized_data) * 0.8)
train_data = tokenized_data[:train_size]
test_data = tokenized_data[train_size:]

# Correct the print statement format and variable names
print(f"Data is split into {len(train_data)} train and {len(test_data)} test set")

# Display first item from both train and test sets
if train_data:  # Check if train_data is not empty
    print("\nFirst item in train data:\n", train_data[0])
if test_data:  # Check if test_data is not empty
    print("\nFirst item in test data:\n", test_data[0])

# Function to count word frequencies in tokenized sentences
def count_words(tokenized_sentences):
    word_count = {}
    for sentence in tokenized_sentences:
        for token in sentence:
            word_count[token] = word_count.get(token, 0) + 1
    return word_count

# Sample tokenized sentences
tokenized_sentences = [['sky', 'is', 'blue', '.'],
                       ['leaves', 'are', 'green', '.'],
                       ['roses', 'are', 'red']]

# Call the function and print the result
word_counts = count_words(tokenized_sentences)
print("\nWord counts:\n", word_counts)

# Get words with frequency >= threshold
def get_word_with_nplus_frequency(tokenized_sentences, count_threshold):
    closed_vocab = []
    word_counts = count_words(tokenized_sentences)
    for word, count in word_counts.items():
        if count >= count_threshold:
            closed_vocab.append(word)
    return closed_vocab

# Define the count threshold
count_threshold = 2

# Call the function and print the result
tmp_closed_vocab = get_word_with_nplus_frequency(tokenized_sentences, count_threshold)
print("\nClosed vocabulary:\n", tmp_closed_vocab)

# Replace OOV words in tokenized sentences
def replace_oov_word(tokenized_sentences, vocabulary, unknown_token='<unk>'):
    replaced_tokenized_sentences = []
    for sentence in tokenized_sentences:
        replaced_sentence = [token if token in vocabulary else unknown_token for token in sentence]
        replaced_tokenized_sentences.append(replaced_sentence)
    return replaced_tokenized_sentences

# Sample tokenized sentences and vocabulary
tokenized_sentences = [["dog", "run"], ["cats", "sleep"]]
vocabulary = ["dog", "sleep"]

# Call the function and display the results
replaced_tokenized_sentences = replace_oov_word(tokenized_sentences, vocabulary)
print("\nReplaced tokenized sentences:\n", replaced_tokenized_sentences)
print("\nVocabulary:\n", vocabulary)

# Preprocess data by replacing OOV words
def preprocess_data(train_data, test_data, count_threshold):
    vocabulary = get_word_with_nplus_frequency(train_data, count_threshold)
    train_data_replaced = replace_oov_word(train_data, vocabulary)
    test_data_replaced = replace_oov_word(test_data, vocabulary)
    return train_data_replaced, test_data_replaced, vocabulary

# Example train and test data
train_data = [['sky', 'is', 'blue', '.'], ['leaves', 'are', 'green', '.']]
test_data = [['roses', 'are', 'red', '.']]

# Preprocess the data
train_data_replaced, test_data_replaced, vocab = preprocess_data(train_data, test_data, 1)

# Display the results
print("\nPreprocessed train data:\n", train_data_replaced)
print("\nPreprocessed test data:\n", test_data_replaced)
print("\nVocabulary:\n", vocab)

# Function to count n-grams
def count_ngrams(data, n, start_token='<s>', end_token='<e>'):
    ngrams = {}
    for sentence in data:
        sentence = [start_token] * n + sentence + [end_token]
        for i in range(len(sentence) - n + 1):
            ngram = tuple(sentence[i:i+n])
            ngrams[ngram] = ngrams.get(ngram, 0) + 1
    return ngrams

# Sample sentences
sentences = [['i', 'like', 'a', 'cat'], ['this', 'dog', 'is', 'like', 'a', 'cat']]

# Count unigrams and bigrams
print("\nUnigrams:\n", count_ngrams(sentences, 1))
print("\nBigrams:\n", count_ngrams(sentences, 2))

# Function to estimate probability of a word given the previous n-gram
def estimate_pro(word, previous_ngram, ngram_counts, n_plus_gram_counts, vocab_size, k=1.0):
    previous_ngram = tuple(previous_ngram)
    previous_ngram_count = ngram_counts.get(previous_ngram, 0)
    n_plus_gram = previous_ngram + (word,)
    n_plus_gram_count = n_plus_gram_counts.get(n_plus_gram, 0)
    probability = (n_plus_gram_count + k) / (previous_ngram_count + k * vocab_size)
    return probability

# Unique words in the sentences
unique_words = list(set(sentences[0] + sentences[1]))

# Get n-gram counts
unigram_counts = count_ngrams(sentences, 1)
bigram_counts = count_ngrams(sentences, 2)

# Example usage of estimate_pro
tmp_prob = estimate_pro("cat", ["a"], unigram_counts, bigram_counts, len(unique_words), k=1)
print("\nProbability of 'cat' given 'a':", tmp_prob)

# Function to estimate probabilities for a vocabulary
def estimate_probabilities(previous_n_gram, n_gram_counts, n_plus1_gram_counts, vocabulary, k=1.0):
    vocabulary = vocabulary + ["<e>", "<unk>"]
    vocab_size = len(vocabulary)
    probabilities = {}
    for word in vocabulary:
        prob = estimate_pro(word, previous_n_gram, n_gram_counts, n_plus1_gram_counts, vocab_size, k)
        probabilities[word] = prob
    return probabilities

# Estimate probabilities
probabilities = estimate_probabilities(["a"], unigram_counts, bigram_counts, unique_words)
print("\nEstimated probabilities:\n", probabilities)

def make_count_matrix(n_plus1_gram_counts, vocab):
    # Ensure special tokens are included in the vocabulary
    vocab = vocab + ["<s>", "<e>", "<unk>"]
    
    # Get unique n-grams from n_plus1_gram_counts
    n_grams = list(set([n_plus1_gram[:-1] for n_plus1_gram in n_plus1_gram_counts.keys()]))
    
    # Create row and column index mappings
    row_index = {n_gram: i for i, n_gram in enumerate(n_grams)}
    
    column_index = {word: j for j, word in enumerate(vocab)}
    
    # Initialize the count matrix
    count_matrix = np.zeros((len(n_grams), len(vocab)))
    
    # Fill the count matrix with the counts from n_plus1_gram_counts
    for n_plus1_gram, count in n_plus1_gram_counts.items():
        n_gram = n_plus1_gram[:-1]
        word = n_plus1_gram[-1]
        
        # Only update the matrix if the n_gram and word exist in the respective indices
        if n_gram in row_index and word in column_index:
            row = row_index[n_gram]
            column = column_index[word]
            count_matrix[row, column] = count

    return count_matrix, row_index, column_index

# Now calling the function should work without errors
count_matrix, row_index, column_index = make_count_matrix(bigram_counts, unique_words)
print("\nCount matrix:\n", count_matrix)
```

```python
import numpy as np
import pandas as pd

# Function to count n-grams
def count_ngrams(sentences, n):
    ngram_counts = {}
    for sentence in sentences:
        for i in range(len(sentence) - n + 1):
            ngram = tuple(sentence[i:i + n])
            if ngram in ngram_counts:
                ngram_counts[ngram] += 1
            else:
                ngram_counts[ngram] = 1
    return ngram_counts

# Function to estimate probabilities
def estimate_pro(previous_n_gram, n_gram_counts, n_plus1_gram_counts, vocab, k=1.0):
    previous_n_gram = tuple(previous_n_gram)
    
    # Count of the n-gram
    n_count = n_gram_counts.get(previous_n_gram, 0)
    
    # Prepare the probabilities dictionary
    probabilities = {}
    
    # Calculate probabilities for the next possible words
    for next_word in vocab:  # Iterate over the vocabulary list
        # Construct n+1-gram
        n_plus1_gram = previous_n_gram + (next_word,)
        
        # Count of the n+1-gram
        n_plus1_count = n_plus1_gram_counts.get(n_plus1_gram, 0)
        
        # Apply Laplace smoothing
        prob = (n_plus1_count + k) / (n_count + k * len(vocab)) if n_count > 0 else 0
        
        probabilities[next_word] = prob
        
    return probabilities

# Suggest a word based on previous tokens
def suggest_a_word(previous_tokens, n_gram_counts, n_plus1_gram_counts, vocab, k=1.0, start_with=None):
    n = len(list(n_gram_counts.keys())[0])  # Find n from the n-gram counts
    previous_n_gram = previous_tokens[-n:]  # Get the previous n-gram from tokens
    
    # Call estimate_pro with the correct arguments
    probabilities = estimate_pro(previous_n_gram, n_gram_counts, n_plus1_gram_counts, vocab, k=k)  
    
    suggestion = None
    max_prob = 0.0
    
    # Loop over the possible words and their probabilities
    for word, prob in probabilities.items():
        # If we want to start with a specific letter
        if start_with is not None:
            if not word.startswith(start_with):
                continue
        
        # Find the word with the highest probability
        if prob > max_prob:
            suggestion = word
            max_prob = prob
    
    return suggestion, max_prob

# Example sentences
sentences = [['i', 'like', 'a', 'cat'],
             ['this', 'dog', 'is', 'like', 'a', 'cat']]

# Get the vocabulary (unique words)
unique_words = list(set(sentences[0] + sentences[1]))

# Get unigram and bigram counts
unigram_counts = count_ngrams(sentences, 1)
bigram_counts = count_ngrams(sentences, 2)

# Test the suggest_a_word function with no start letter
previous_tokens = ['i', 'like']
tmp_suggest1 = suggest_a_word(previous_tokens, unigram_counts, bigram_counts, unique_words, k=1.0)
print("Suggested word:", tmp_suggest1[0], "with probability:", tmp_suggest1[1])

print()

# Test the suggest_a_word function with a start letter
tmp_starts_with = 'c'
tmp_suggest2 = suggest_a_word(previous_tokens, unigram_counts, bigram_counts, unique_words, k=1.0, start_with=tmp_starts_with)
print("Suggested word starting with 'c':", tmp_suggest2[0], "with probability:", tmp_suggest2[1])
```
