```python
import os
import shutil

# Define paths
source_folder_1 = r'D:\merge_news\fake_news'
source_folder_2 = r'D:\merge_news\real_news'
destination_folder = r'D:\merge_news\merged_news'  # Create a new destination folder

# Create destination folder if it doesn't exist
if not os.path.exists(destination_folder):
    os.makedirs(destination_folder)

# Function to merge files
def merge_folders(source_folder, destination_folder):
    for filename in os.listdir(source_folder):
        source_file = os.path.join(source_folder, filename)
        dest_file = os.path.join(destination_folder, filename)
        
        # Rename the file if it already exists to avoid overwriting
        if os.path.exists(dest_file):
            base, ext = os.path.splitext(filename)
            i = 1
            while os.path.exists(dest_file):
                new_filename = f"{base}_{i}{ext}"
                dest_file = os.path.join(destination_folder, new_filename)
                i += 1
                
        # Copy the file
        shutil.copy2(source_file, dest_file)

# Merge real_news and fake_news folders
merge_folders(source_folder_1, destination_folder)
merge_folders(source_folder_2, destination_folder)

print("Folders merged successfully!")
```

```python
import pandas as pd

# Define file paths
file_1 = r'D:\merge_news\marathi_fake_news_dataset.xlsx'
file_2 = r'D:\merge_news\marathi_real_news_dataset.xlsx'
output_file = r'D:\merge_news\merged_marathi_news_dataset.xlsx'

# Load the datasets
df_fake_news = pd.read_excel(file_1)
df_real_news = pd.read_excel(file_2)

# Merge the datasets by concatenating them
merged_df = pd.concat([df_fake_news, df_real_news], ignore_index=True)

# Save the merged dataset to a new Excel file
merged_df.to_excel(output_file, index=False)

print(f"Files merged successfully into {output_file}")
```
