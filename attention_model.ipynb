```python
import numpy as np
def softmax(x,axis=0):
    return np.exp(x)/np.expand_dims(np.sum(np.exp(x),axis=axis),axis)
```

```python
import numpy as np

hidden_size = 16
attention_size = 10
input_length = 5

np.random.seed(42)
encoder_state = np.random.randn(input_length, hidden_size)
decoder_state = np.random.randn(1, hidden_size)
layer_1 = np.random.randn(2 * hidden_size, attention_size)
layer_2 = np.random.randn(attention_size, 1)

def alignment(encoder_states, decoder_state):
    # Repeat decoder_state to match the encoder_states shape
    decoder_state_repeated = np.repeat(decoder_state, input_length, axis=0)
    
    # Concatenate encoder and decoder states along the last dimension
    inputs = np.concatenate([encoder_states, decoder_state_repeated], axis=1)
    assert inputs.shape == (input_length, 2 * hidden_size), f"Shape mismatch: {inputs.shape}"
    
    # First layer transformation with tanh activation
    activations = np.tanh(np.matmul(inputs, layer_1))
    assert activations.shape == (input_length, attention_size), f"Shape mismatch: {activations.shape}"
    
    # Second layer transformation to get scores
    scores = np.matmul(activations, layer_2).flatten()
    assert scores.shape == (input_length,), f"Shape mismatch: {scores.shape}"
    
    # Apply softmax to get attention weights
    attention_weights = np.exp(scores) / np.sum(np.exp(scores))
    
    return attention_weights

# Compute attention weights
attention_weights = alignment(encoder_state, decoder_state)
print("Attention Weights:", attention_weights)
```

```python
scores = alignment(encoder_state, decoder_state)
print(scores)
```

```python
def attention(encoder_states, decoder_state):
    # Get alignment scores
    scores = alignment(encoder_states, decoder_state)
    
    # Apply softmax to get attention weights
    weights = softmax(scores)
    
    # Calculate weighted sum of encoder states
    weighted_scores = [state * weight for state, weight in zip(encoder_states, weights)]
    context = np.sum(weighted_scores, axis=0)  # Summing to create the context vector
    
    return context

# Compute context vector
context_vector = attention(encoder_state, decoder_state)
print("Context Vector:", context_vector)
```
